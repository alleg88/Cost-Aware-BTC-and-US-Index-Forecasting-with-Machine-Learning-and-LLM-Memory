# Run from ZIP in Google Colab
Select **Runtime → Change runtime type → CPU**, choose a mode below and select **Runtime → Run all**. When **Choose files** appears, select **cost-aware-market-forecasting-code.zip**, not the notebook or run_zip.py. For **Rebuild results**, select both the code and source ZIPs together. Wait for the upload to finish; setup then continues automatically. No GitHub token or LLM key is needed.

**Check installation** works with the code ZIP alone; it verifies the code, environment and tests without recalculating results. **Rebuild results** requires both ZIPs and also refits models and updates the result notebooks. Full rebuilding can exceed Colab's memory, disk or session limits.

Runs in Colab's own Python; only missing packages are installed. Rerunning cells reuses the uploaded ZIPs. You can add the source ZIP later for a rebuild. Changing or deleting the runtime removes its temporary files; download rebuilt notebooks before ending the session. Colab's library versions can differ from the fixed local environment, so recalculated values may differ from the saved results.

```python
#@title 1. Choose a mode
mode = "Check installation" #@param ["Check installation", "Rebuild results"]
```

```python
#@title 2. Upload once and run
from pathlib import Path
from zipfile import ZipFile
archives = Path("/content")
required = ["cost-aware-market-forecasting-code.zip"]
if mode == "Rebuild results":
    required.append("cost-aware-market-forecasting-sources.zip")
missing = [name for name in required if not (archives / name).is_file()]
if missing:
    from google.colab import files
    from IPython.display import clear_output
    print("Choose the ZIP archive(s): " + ", ".join(missing), flush=True)
    uploaded = files.upload(target_dir=str(archives))
    del uploaded
    clear_output(wait=False)
missing = [name for name in required if not (archives / name).is_file()]
if missing:
    raise ValueError("Missing: " + ", ".join(missing) + ". Run this cell again and choose the ZIP archive(s), not the notebook or run_zip.py.")
print("Using ZIPs saved in this runtime; no further upload is needed.", flush=True)
code_zip = archives / "cost-aware-market-forecasting-code.zip"
with ZipFile(code_zip) as package:
    source = package.read("cost-aware-market-forecasting/run_zip.py")
launcher = {"__name__": "project_zip_launcher"}
exec(compile(source, str(code_zip) + "/run_zip.py", "exec"), launcher)
project = launcher["run"](mode, archive_dir=archives, native=True)
```
